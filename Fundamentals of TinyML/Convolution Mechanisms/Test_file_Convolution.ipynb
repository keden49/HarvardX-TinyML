# Visualizing Grid system

```python
import matplotlib.pyplot as plt
import numpy as np

# Define image dimensions for illustration
image_height = 10
image_width = 10

plt.figure(figsize=(7, 7))
ax = plt.gca()

# Set limits for the plot, extending slightly beyond the image dimensions for clarity
ax.set_xlim(-0.5, image_width - 0.5)
ax.set_ylim(-0.5, image_height - 0.5)

# Invert the y-axis to represent (0,0) at the top-left, and x increasing downwards
ax.invert_yaxis()

# Set ticks to align with pixel centers for a clearer grid representation
ax.set_xticks(np.arange(0, image_width, 1))
ax.set_yticks(np.arange(0, image_height, 1))

# Add grid lines
ax.grid(True, which='both', color='lightgray', linestyle='-', linewidth=0.5)

# Label axes according to the description
ax.set_xlabel('Y-axis (Columns / Width)', fontsize=12)
ax.set_ylabel('X-axis (Rows / Height)', fontsize=12)

# Annotate the origin (0,0)
ax.plot(0, 0, 'ro') # Red dot for origin
ax.text(0.2, 0.2, '(0, 0) Origin', color='red', fontsize=10, verticalalignment='top', horizontalalignment='left')

# Add arrows to indicate axis directions
# X-axis (Rows / Height) - downwards
ax.annotate('', xy=(0, image_height * 0.7), xytext=(0, 0.5), arrowprops=dict(facecolor='blue', shrink=0.05, width=1, headwidth=8), zorder=3)
ax.text(0.5, image_height * 0.7, 'X-axis (Rows/Height) increases', color='blue', rotation=90, verticalalignment='center', horizontalalignment='left', fontsize=9)

# Y-axis (Columns / Width) - rightwards
ax.annotate('', xy=(image_width * 0.7, 0), xytext=(0.5, 0), arrowprops=dict(facecolor='green', shrink=0.05, width=1, headwidth=8), zorder=3)
ax.text(image_width * 0.7, 0.5, 'Y-axis (Cols/Width) increases', color='green', verticalalignment='bottom', horizontalalignment='center', fontsize=9)

# Title
plt.title('Image Coordinate System Visualization', fontsize=14)

# Ensure tight layout and display plot
plt.tight_layout()
plt.show()
```
